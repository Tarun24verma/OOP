## Question 3 — Flexible Bill Calculator

**Level:** Easy to Medium  
**Concepts:** `*args`, `**kwargs`, variable-length arguments

Write a function named `calculate_bill` that:

* Accepts any number of item prices using `*args`.
* Accepts optional settings such as `discount` and `tax` using `**kwargs`.
* Calculates the subtotal.
* Applies the discount percentage first.
* Applies the tax percentage after the discount.
* Returns the final bill amount.

Test the function with at least four prices, one discount percentage, and one tax percentage.

\---

````python
def calculate_bill(*prices,**charges):
    sub_total=sum(prices)
    discount=charges.get("discount",0)
    tax=charges.get("tax",0)
    discounted=sub_total-(sub_total*discount/100)
    return discounted+(discounted*tax/100)
````

````python
print(calculate_bill(120, 250, 80, 150, discount=10,tax=18))
````

Display the original marks, updated marks, passing marks, and final total.

````python
nums=[96,97,95,80,86,50,35,30]
print(nums)
graced=list(map(lambda x: x+5 if x+5<=100 else x, nums))
print(graced)
filtered_grace=list(filter(lambda x: x>=40, graced))
print(filtered_grace)
from functools import reduce
total=reduce(lambda x,y:x+y, filtered_grace)
print(total)
````

## Question 6 — Function Call Logger Decorator

**Level:** Medium  
**Concepts:** Decorators, `*args`, `**kwargs`, `functools.wraps`

Create a decorator named `log_call` that:

* Displays the name of the decorated function.
* Displays all positional and keyword arguments received.
* Calls the original function.
* Displays the returned result.
* Preserves the original function metadata using `functools.wraps`.

Apply the decorator to a function named `calculate_average` that accepts any number of marks and returns their average.

The function must raise a suitable error when no marks are supplied.

\---

````python
from functools import wraps

def log_call(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        print(f"Function Name: {func.__name__}")
        print(f"Positional Arguments: {args}")
        print(f"Keyword Arguments: {kwargs}")

        result = func(*args, **kwargs)

        print(f"Returned Result: {result}")
        return result

    return wrapper


@log_call
def calculate_average(*marks):
    if not marks:
        raise ValueError("At least one mark must be provided.")

    return sum(marks) / len(marks)
````

````python
print(calculate_average(80, 90, 100))
````

## Question 7 — Permission Decorator with Arguments

**Level:** Medium to High  
**Concepts:** Decorators with arguments, nested functions, validation

Create a decorator named `require_role` that accepts a required role as its argument.

The decorator must check the `role` keyword argument passed to the decorated function:

* If the role matches the required role, allow the function to execute.
* Otherwise, return or display an access-denied message.

Apply the decorator to a function named `delete_student_record` that accepts a student ID and simulates deleting the record.

Test the function using both an authorized role and an unauthorized role.

\---

````python
def require_role(required_role):
    def decorator(func):
        @wraps(func)
        def wrapper(*args, **kwargs):
            role = kwargs.get("role")

            if role == required_role:
                return func(*args, **kwargs)
            else:
                return f"Access Denied! '{role}' does not have permission."
        return wrapper
    return decorator


@require_role("admin")
def delete_student_record(student_id, role=None):
    return f"Student record with ID {student_id} has been deleted."
````

````python
print(delete_student_record(101, role="admin"))
````

````python
print(delete_student_record(102, role="teacher"))
````
