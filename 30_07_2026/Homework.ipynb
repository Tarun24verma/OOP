A keyword-only argument named `include_status`, with a default value of `True`
The function must return a formatted report. When `include_status` is `True`, include `Pass` for marks of 40 or above and `Fail` otherwise.
Demonstrate the function using both positional and keyword arguments.


```python
def create_student_report(student_name='Not Assigned', marks='Not Assigned', grade='Not Assigned',*,include_status=True):
    print(f"Name: {student_name}\nMarks:{marks}\nGrade: {grade}")
    if include_status:
        if marks>=40:
            print("Pass")
        else: print("Fail")
```

Question 3 — Flexible Bill Calculator
Level: Easy to Medium
Concepts: \*args, \*\*kwargs, variable-length arguments

Write a function named calculate\_bill that:

Accepts any number of item prices using \*args.
Accepts optional settings such as discount and tax using \*\*kwargs.
Calculates the subtotal.
Applies the discount percentage first.
Applies the tax percentage after the discount.
Returns the final bill amount.
Test the function with at least four prices, one discount percentage, and one tax percentage.

---

```python
def calculate_bill(*prices,**charges):
    sub_total=sum(prices)
    discount=charges.get("discount",0)
    tax=charges.get("tax",0)
    discounted=sub_total-(sub_total*discount/100)
    return discounted+(discounted*tax/100)
```

```python
print(calculate_bill(120, 250, 80, 150, discount=10,tax=18))
```

Question 5 — Marks Processing with Functional Tools
Level: Medium
Concepts: Lambda functions, map, filter, reduce

Given a list of student marks:

Use map and a lambda function to add five grace marks to every score without allowing any score to exceed 100.
Use filter and a lambda function to keep only marks of 40 or above.
Use reduce to calculate the total of the passing marks.
Display the original marks, updated marks, passing marks, and final total.

---

```python
nums=[96,97,95,80,86,50,35,30]
print(nums)
graced=list(map(lambda x: x+5 if x+5<=100 else x, nums))
print(graced)
filtered_grace=list(filter(lambda x: x>=40, graced))
print(filtered_grace)
from functools import reduce
total=reduce(lambda x,y:x+y, filtered_grace)
print(total)
```
